# Een `Date` klasse

Data

- dag
- maand
- jaar

````python
class Date:
    """Een eigen datatype: een klasse die datums opslaat en bewerkt."""

    def __init__(self, day, month, year):
        """De constructor voor objecten van het type Date."""
        self.day = day
        self.month = month
        self.year = year

    def __repr__(self):
        """Geeft een string die het object self voorstelt."""
        return f"{self.day:02d}/{self.month:02d}/{self.year:04d}"

    def is_leap_year(self):
        """Geeft True als self, het aanroepende object, in een schrikkeljaar valt."""
        if self.year % 400 == 0:
            return True
        if self.year % 100 == 0:
            return False
        if self.year % 4 == 0:
            return True
        return False
````

### `self`

`self` is het object waarvoor de methode wordt aangeroepen. Bij `wd.is_leap_year()` is `self` het object `wd`. Net als bij de constructor vult Python `self` zelf in: `delay(self, num_years)` uit het vorige college roep je aan als `student_b.delay(1)`, en dan is `self` het object `student_b`.

Misschien heb je al kennisgemaakt met andere programmeertalen en zal je herkennen dat `self` overeenkomt met wat je kent als `this` in deze talen (bijvoorbeeld in Java of C++). Guido van Rossum, de maker van Python, heeft uitgelegd waarom hij niet voor `this` koos, en noemt het een daad van speels verzet.

Het is hier misschien ook van belang om stil te staan bij het feit dat programmeertalen niet zomaar zijn ontstaan, het is mensenwerk en wij bouwen weer voort op het werk van anderen!

## Waarde en identiteit

Maak twee datums voor dezelfde dag, en druk ze allebei af:

````python
wd = Date(11, 12, 2013)
wd2 = Date(11, 12, 2013)
print(wd)
print(wd2)
````

Vergelijk ze nu met `==`:

````python
wd == wd2
````

Twee keer dezelfde datum, en toch zegt Python `False`. `wd` en `wd2` zijn
twee objecten. Elke aanroep `Date(...)` maakt een nieuw object, ook als de
argumenten dezelfde zijn. De twee objecten hebben dezelfde dag, maand en jaar,
maar het blijven er twee.

Het kan ook andersom: één object met twee namen. Neem de klasse `Student` uit het
vorige college, de versie met `_year`:

````python
class Student:
    """Een student met een naam en een startjaar."""

    def __init__(self, name, year):
        """Maak een student met de gegeven naam en het gegeven startjaar."""
        self.name = name
        self._year = year

    def __repr__(self):
        """Geeft de student als string, om af te drukken."""
        return "naam: " + self.name + ", startjaar: " + str(self._year)

    def delay(self, num_years):
        """Stel de start van de studie num_years jaar uit, en nooit terug."""
        if num_years > 0:
            self._year += num_years

    @property
    def year(self):
        """Het startjaar, alleen om te lezen."""
        return self._year
````

Maak `student_b` en geef hem een tweede naam, `jasper`. Stel dan de studie uit
via `jasper`, en druk `student_b` af:

````python
student_b = Student("Jasper Klein", 2021)
jasper = student_b
jasper.delay(1)
print(student_b)
````

Je hebt de studie niet via `student_b` uitgesteld, en toch is zijn startjaar
veranderd. `jasper = student_b` maakt geen nieuwe student. Een variabele bevat
het object niet zelf, maar een **verwijzing** ernaar: ze wijst het object aan.
Na `jasper = student_b` wijzen beide namen hetzelfde object aan, dus wat je via
de ene naam verandert, zie je via de andere.

| Code | Wat er gebeurt |
|---|---|
| `jasper = student_b` | geen nieuw object: `jasper` is een tweede naam voor hetzelfde object |
| `wd2 = Date(11, 12, 2013)` | een nieuw object, met dezelfde datum als `wd` |

De begrippen op een rij:

| Begrip | Wat het is | Hierboven |
|---|---|---|
| **waarde** | wat er in een object staat: de waarden van zijn attributen | `wd` en `wd2` hebben dezelfde waarde |
| **identiteit** | welk object het is | `jasper` en `student_b` hebben dezelfde identiteit: het is één object |
| **verwijzing** | wat een variabele bevat: ze wijst een object aan | `jasper` en `student_b` bevatten een verwijzing naar hetzelfde object |

In Programmeren 1 zag je al dat twee namen dezelfde lijst kunnen aanwijzen, in
het college [More on Lists](/lectures/7a_lists_advanced.ipynb#lists-en-shallow-copy).
Bij je eigen objecten werkt dat net zo.

### `is` en `==`

Met `is` vraag je naar de identiteit. `a is b` geeft `True` als `a` en `b`
hetzelfde object aanwijzen, en anders `False`:

````python
print(jasper is student_b)
print(wd is wd2)
````

Het omgekeerde is `is not`: `wd is not wd2` geeft `True`.

En `==`? Bij je eigen objecten, van een klasse die je zelf schrijft, vergelijkt
`==` of het hetzelfde object is: de identiteit, net als `is`. Daarom gaf
`wd == wd2` hierboven `False`. Dat geldt alleen voor je eigen objecten. Bij
getallen, strings en lijsten vergelijkt `==` de waarde, zoals je gewend bent.

Wil je bij je eigen objecten weten of twee namen hetzelfde object aanwijzen,
schrijf dan `is`: dat zegt wat je bedoelt. Wil je de waarde vergelijken, dan
heb je een eigen methode nodig. In de basisopgave krijg je daarvoor de methode
`equals`.

`None` bestaat maar één keer. Of een variabele `None` aanwijst, vraag je daarom
ook met `is`:

````python
best = None
print(best is None)
best = student_b
print(best is None)
````
