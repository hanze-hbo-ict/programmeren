# Een `Date` klasse

Data

- dag
- maand
- jaar

````python
class Date:
    """Een eigen datatype: een klasse die datums opslaat en bewerkt."""

    def __init__(self, day, month, year):
        """De constructor voor objecten van het type Date."""
        self.day = day
        self.month = month
        self.year = year

    def __repr__(self):
        """Geeft een string die het object self voorstelt."""
        return f"{self.day:02d}/{self.month:02d}/{self.year:04d}"

    def is_leap_year(self):
        """Geeft True als self, het aanroepende object, in een schrikkeljaar valt."""
        if self.year % 400 == 0:
            return True
        if self.year % 100 == 0:
            return False
        if self.year % 4 == 0:
            return True
        return False
````

Let op de dubbele punt: alles wat volgt, springt in, omdat het bij de klasse hoort.

```python

```

Dit is de *constructor* voor `Date`-objecten. Hier krijgen de attributen van een nieuw object hun beginwaarde, uit de argumenten.

De attributen zijn de gegevens die elk `Date`-object heeft, elk object met zijn eigen waarden.

Dit is de string die een `Date`-object voorstelt: zo weet Python hoe het object afgedrukt moet worden.

Waarom wordt `self` gebruikt en niet `d`?

### `self`

`self` is het object waarvoor de methode wordt aangeroepen. Bij `wd.is_leap_year()` is `self` het object `wd`.

Misschien heb je al kennisgemaakt met andere programmeertalen en zal je herkennen dat `self` overeenkomt met wat je kent als `this` in deze talen (bijvoorbeeld in Java of C++). Guido van Rossum geeft hier antwoord op de vraag waarom niet voor `this` is gekozen, en legt het uit als een daad van speels verzet.

Het is hier misschien ook van belang om stil te staan bij het feit dat programmeertalen niet zomaar zijn ontstaan, het is mensenwerk en wij bouwen weer voort op het werk van anderen!
