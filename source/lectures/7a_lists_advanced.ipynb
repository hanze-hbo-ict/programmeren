## Eerst directe mutatie

Een lijst is mutable: `bord[rij][kolom] = waarde` verandert het bestaande bord. Dat is handig wanneer één cel moet worden bijgewerkt. De volgorde is bewust: eerst het vermogen, daarna de valkuil.

```python
bord = [[0, 0], [0, 0]]
bord[0][1] = 1
assert bord == [[0, 1], [0, 0]]
```

## Aliasing en kopiëren

Bij `ander = bord` krijgen twee namen dezelfde lijst. Een wijziging via de ene naam zie je via de andere. Een onafhankelijke rasterkopie maakt voor iedere rij een nieuwe lijst:

```python
bord = [[0, 0], [0, 0]]
kopie = []
for rij in bord:
    kopie = kopie + [rij[:]]
kopie[0][0] = 1
assert bord == [[0, 0], [0, 0]]
```

## Deelprobleem en functiecompositie

Een algoritme wordt hanteerbaar als elke functie één deelprobleem oplost. In de basis tellen `count_neighbors` en `next_cell` eerst afzonderlijk; `next_generation` composeert die functies. Schrijf bij iedere functie de invoer, uitvoer en een kleine assertion voordat je verdergaat.

## Tuple en uitpakken

Een tuple kan twee samenhangende waarden teruggeven. Deze week is één vorm genoeg:

```python
def midden_van(bord):
    return (len(bord) // 2, len(bord[0]) // 2)

rij, kolom = midden_van([[0, 0, 0], [0, 1, 0], [0, 0, 0]])
assert (rij, kolom) == (1, 1)
```
