Een functie krijgt argumenten mee: getallen, strings, lijsten. In dit college

## Sorteren, het grootste en het kleinste

Python heeft drie ingebouwde functies die je deze week veel gebruikt:

| Functie | Geeft terug |
|---|---|
| `sorted(L)` | een nieuwe lijst met de elementen van `L`, van klein naar groot |
| `max(L)` | het grootste element van `L` |
| `min(L)` | het kleinste element van `L` |

`max` ken je uit het [vorige college](/lectures/11a_use_it_or_lose_it), met twee
argumenten. Je kunt haar ook één lijst geven.

````python
numbers = [5, 2, 8]
print(sorted(numbers))
print(max(numbers))
print(min(numbers))
````

Strings vergelijkt Python op alfabet: `"appel"` komt vóór `"peer"`. `sorted`
geeft een nieuwe lijst terug en laat de oude lijst staan.

````python
fruit = ["peer", "appel", "kiwi", "banaan"]
print(sorted(fruit))
print(max(fruit))
print(fruit)
````

## Een functie meegeven met `key=`

Stel dat je de vruchten niet op alfabet wilt sorteren, maar op de lengte van hun
naam. Dan geef je `sorted` een functie mee die bij elk element zegt waarop
gesorteerd wordt. Dat doe je met `key=`:

````python
print(sorted(fruit, key=len))
print(max(fruit, key=len))
print(min(fruit, key=len))
````

Let op: `len` staat er zonder haakjes. Je roept `len` hier niet zelf aan, je
geeft de functie mee aan `sorted`. `sorted` roept `len` daarna aan bij elk
element en sorteert op wat er terugkomt: 4 voor `"peer"`, 5 voor `"appel"`,
enzovoort. Bij gelijke lengte blijft de oorspronkelijke volgorde staan; daarom
staat `"peer"` vóór `"kiwi"`.

Dat heet een **functie als argument**: je geeft een functie mee aan een andere
functie, die haar zelf aanroept. De functie achter `key=` mag ook een functie zijn
die je zelf schrijft:

````python
def last_letter(s):
    """Geeft de laatste letter van s."""
    return s[-1]


print(sorted(fruit, key=last_letter))
````

Ook een methode kun je meegeven. Bij een dictionary geeft `totals.get(name)` de
waarde bij de sleutel `name`. Zonder haakjes is `totals.get` de functie zelf.
Dit is de ranglijst uit de vooruitblik van week 2:

````python
totals = {"Anne": 5, "Bram": 7, "Cas": 4, "Daan": 1}
print(sorted(totals, key=totals.get))
print(max(totals, key=totals.get))
````

`sorted` loopt de sleutels van de dictionary langs, de namen, en sorteert ze op
wat `totals.get` bij elke naam teruggeeft: de punten.

## `lambda`: een functie zonder naam

Vaak is de functie achter `key=` zo klein dat ze geen eigen `def` en geen naam
verdient. Dan schrijf je haar ter plekke op, met `lambda`:

```python
lambda s: s[-1]
```

Dit is een functie met de parameter `s`, die `s[-1]` teruggeeft. Ze doet dus
hetzelfde als `last_letter`. Tussen `lambda` en de dubbele punt staat de
parameter; na de dubbele punt staat wat de functie teruggeeft. Een `return`
schrijf je er niet bij. Zo'n functie heet een **lambda-functie**.

````python
print(sorted(fruit, key=lambda s: s[-1]))
````

Een lambda-functie is vooral handig bij een lijst van lijsten. Hier staan de
spelers als paren: een naam en een aantal punten.

````python
players = [["Anne", 5], ["Bram", 7], ["Cas", 4], ["Daan", 1]]

print(max(players, key=lambda p: p[1]))
print(sorted(players, key=lambda p: p[1]))
print(sorted(players, key=lambda p: -p[1]))
````

`lambda p: p[1]` geeft van elke speler de punten. Daarop zoekt `max` de winnaar,
en daarop sorteert `sorted`, van weinig naar veel punten. Wil je de ranglijst van
veel naar weinig, sorteer dan op min de punten: `-p[1]`.

## Zelf een functie met een functieparameter

Ook in een eigen functie kan een parameter een functie zijn. In de basisopgave
van [vorige week](/problems/10_basis) schreef je `best_word`: het woord met de
hoogste Scrabble-score, met dezelfde vorm als `largest`. `best` hieronder doet
hetzelfde, maar de score is nu een parameter. Bij `score(L[0])` roept `best` de
functie aan die je hebt meegegeven.

````python
def best(L, score):
    """Geeft het element uit L met de hoogste score; L is niet leeg."""
    if len(L) == 1:  # basisgeval
        return L[0]

    rest = best(L[1:], score)  # het beste element in de rest van L

    if score(L[0]) >= score(rest):
        return L[0]
    else:
        return rest


assert best(fruit, len) == "banaan"
assert best(players, lambda p: p[1]) == ["Bram", 7]
assert best([3, -8, 5], abs) == -8
````

Dezelfde `best` zoekt de langste naam, de speler met de meeste punten, of het
getal dat het verst van 0 ligt. Alleen de functie die je meegeeft, verschilt.
Zo werkt `max` met `key=` ook.

## Terug naar de knapzak

`subset` uit het vorige college gaf de grootste som, maar niet welke spullen
daarbij horen. Laat *use it* en *lose it* nu lijsten met spullen teruggeven in
plaats van sommen. Dan kies je de lijst met de grootste som:
`max(use_it, lose_it, key=sum)`. `sum` ken je uit week 2.

````python
def subset_items(capacity, items):
    """Geeft de items met de grootste som die niet boven capacity uitkomt."""
    if items == []:  # basisgeval: geen items meer
        return []

    first = items[0]
    rest = items[1:]

    if first > capacity:  # recursief geval: het eerste item past niet
        return subset_items(capacity, rest)
    else:  # recursief geval: probeer beide keuzes
        use_it = [first] + subset_items(capacity - first, rest)
        lose_it = subset_items(capacity, rest)
        return max(use_it, lose_it, key=sum)


assert subset_items(10, [8, 4, 6]) == [4, 6]
assert subset_items(42, [30, 10, 45, 5]) == [30, 10]
assert subset_items(10, []) == []
````

De vorm is die van `subset`. Twee dingen zijn anders: *use it* plakt
`[first]` voor de lijst van de rest in plaats van `first` op te tellen, en `max`
vergelijkt de twee lijsten op hun som.
