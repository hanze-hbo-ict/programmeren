### Faculteit met `for`

````python
def fac(n):
    result = 1                      # not the result yet!

    for number in range(1, n + 1):  # range start at 0, add one!
        result *= number            # result = result * i

    return result                   # return the result
````

````python
fac(5)
````

### Ontsnappen

````python
import random

guess = 0  # starting value, not the final or desired value!

while guess != 42:  # test to see if we keep looping
    print('Help! Let me out!')
    guess = random.choice([41, 42, 43])  # watch out for infinite loops!

print('At last!')  # after the loop ends
````

### Verschillen

Wat zijn de verschillen in ontwerp tussen deze twee Python lussen?

`for` - begrensde herhaling

Het aantal herhalingen ligt vast voordat de lus begint: de lengte van de lijst, of het bereik van `range`.


`while` - onbegrensde herhaling

Het aantal herhalingen ligt niet vast. Jij schrijft de conditie die de lus laat stoppen, en jij zorgt dat die ooit waar wordt.

## Het lusrecept

Bijna elke lus in deze week doet hetzelfde: iets verzamelen terwijl je ergens langsloopt.

Vijf vragen brengen je van het probleem naar de code. Beantwoord ze eerst, schrijf dan pas.

1. **Wat verzamel je, en wat is de startwaarde?** Die variabele heet de *verzamelvariabele*.
2. **Wat loop je langs?** Daaruit volgt `for` of `while`. Bij een `for` kies je hier ook tussen de twee typen hierboven: op element of op index.
3. **Wat gebeurt er per stap met wat je verzamelt?**
4. **Wanneer is het klaar?** Bij een begrensde lus ligt dat vast. Bij een onbegrensde lus schrijf je het zelf.
5. **Wat geef je terug, en waar staat die regel?**

Vraag 4 is de vraag die er tot nu toe niet stond. Bij een `for` beantwoordt Python hem voor je: de lijst raakt op. Bij een `while` niet. Onbegrensd wil zeggen dat de grens van de schrijver moet komen, en dat is precies wat vraag 4 vraagt.

### Voorbeeld: `total(L)`

De som van de getallen in een lijst.

1. Verzamelen: `result`, startwaarde `0`
2. Langs: de elementen van `L`, dus een `for` op element
3. Per stap: `result` erbij optellen
4. Klaar: als `L` op is
5. Terug: `return result`, **na** de lus

````python
def total(L):
    """Geeft de som van de getallen in de lijst L."""
    result = 0  # 1. verzamelen, startwaarde

    for number in L:  # 2. langs de elementen  4. klaar als L op is
        result = result + number  # 3. per stap erbij

    return result  # 5. na de lus


assert total([1, 2, 3]) == 6
assert total([-4, 4]) == 0
assert total([]) == 0
````

`result` is de naam die het practicum ook gebruikt en die in opdracht 2 hieronder terugkomt. Eén naam voor één idee: dit is de verzamelvariabele.

### Voorbeeld: `power_of_two(limit)`

De kleinste macht van 2 die minstens `limit` is.

1. Verzamelen: `result`, startwaarde `1`
2. Langs: geen bestaande reeks, dus een `while`
3. Per stap: `result` verdubbelen
4. Klaar: zodra `result` minstens `limit` is. Elke stap verdubbelt `result`, dus dat gebeurt ook echt
5. Terug: `return result`, **na** de lus

````python
def power_of_two(limit):
    """Geeft de kleinste macht van 2 die minstens limit is."""
    result = 1  # 1. verzamelen, startwaarde

    while result < limit:  # 2. een while  4. klaar zodra result groot genoeg is
        result = result * 2  # 3. per stap verdubbelen

    return result  # 5. na de lus


assert power_of_two(1) == 1
assert power_of_two(42) == 64
assert power_of_two(64) == 64
````

Hier doet vraag 4 het werk. Deze lus stopt omdat `result` elke stap groter wordt en `limit` dus een keer passeert. Haal die vermenigvuldiging weg en de lus stopt nooit. Dat is geen eigenschap van `while`, dat is een fout in stap 3.
