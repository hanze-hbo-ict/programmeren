# List comprehensions

## Een lus die een nieuwe lijst verzamelt

In [week 4 van Programmeren 1](/course/week_4) schreef je lussen met het
[lusrecept](/lectures/4a_lussen.ipynb#het-lusrecept). Veel van die lussen doen
hetzelfde: ze lopen een reeks langs en verzamelen een **nieuwe lijst**. Deze
functie houdt de even getallen uit een lijst over:

````python
def only_evens(L):
    """Geeft een lijst met de even getallen uit L."""
    result = []

    for x in L:
        if x % 2 == 0:
            result.append(x)

    return result


assert only_evens([0, 1, 2, 3, 4]) == [0, 2, 4]
assert only_evens([]) == []
````

Met het lusrecept erbij:

| Vraag | Bij `only_evens` |
|---|---|
| 1. Wat verzamel je, en wat is de startwaarde? | een nieuwe lijst `result`, startwaarde `[]` |
| 2. Wat loop je langs? | de elementen van `L` |
| 3. Wat gebeurt er per stap met wat je verzamelt? | is `x` even, dan komt `x` erbij |
| 4. Wanneer is het klaar? | als `L` op is |
| 5. Wat geef je terug, en waar staat die regel? | `result`, na de lus |

Bij zo'n lus zijn vraag 1, 4 en 5 altijd hetzelfde. Je begint met een lege
lijst, de `for` stopt vanzelf als de reeks op is, en na de lus geef je de lijst
terug. Alleen vraag 2 en 3 zeggen iets over dít probleem. Voor precies dit
recept heeft Python een kortere schrijfwijze.

## De list comprehension

````python
def only_evens(L):
    """Geeft een lijst met de even getallen uit L."""
    return [x for x in L if x % 2 == 0]


assert only_evens([0, 1, 2, 3, 4]) == [0, 2, 4]
assert only_evens([]) == []
````

Dit heet een **list comprehension**: een lijst die je in één uitdrukking
opbouwt uit een andere reeks. De term is Engels en blijft dat.

Lees hem in drie delen:

| Deel | Bij `only_evens` | In het lusrecept |
|---|---|---|
| wat er in de nieuwe lijst komt | `x` | vraag 3: wat erbij komt |
| wat je langsloopt | `for x in L` | vraag 2 |
| welke elementen meedoen | `if x % 2 == 0` | vraag 3: óf het erbij komt |

De vierkante haken zeggen dat het resultaat een lijst is. De lege startlijst, de
`append` en de `return` na de lus staan er niet meer: die doet de list
comprehension voor je. Het derde deel heet ook wel het **filter**. Als schema:

```text
[ wat erin komt   for element in reeks   if conditie ]
```

Beide versies van `only_evens` zijn goed en geven dezelfde lijst terug. Wanneer
je beter de lus houdt, zie je in het
[tweede college van deze week](/lectures/9b_dict_en_set_comprehensions).

### Zonder filter

Het filter mag weg. Dan komt er voor elk element iets in de nieuwe lijst. Wat
erin komt hoeft niet het element zelf te zijn; het mag ook iets zijn wat je
ermee uitrekent. De lengte van elk woord:

````python
words = ["de", "vuurtoren", "draait"]
[len(word) for word in words]
````

### Langs `range` en langs een string

Wat je langsloopt, mag elke reeks zijn waar een `for` langs kan. `range` levert
getallen. De eerste vijf veelvouden van 5:

````python
[x * 5 for x in range(1, 6)]
````

Een string levert tekens. De klinkers in een woord:

````python
[c for c in "vuurtoren" if c in "aeiou"]
````

Soms heb je het element zelf niet nodig, alleen het aantal keer. Noem het dan
`_`, zoals in week 4: de naam voor een variabele die je bewust niet gebruikt.
Een lijst met vier nullen:

````python
[0 for _ in range(4)]
````

## Positie en element tegelijk: `enumerate`

In week 4 liep je soms langs de index in plaats van langs de elementen, met
`range(len(...))`:

````python
words = ["spam", "taart", "spam"]

for ix in range(len(words)):
    print(ix, words[ix])
````

Vaak heb je allebei nodig: de positie en het element. `enumerate` geeft je ze
tegelijk, als paren:

````python
list(enumerate(words))
````

Zo'n paar pak je uit zoals de paren van `.items()` in week 1. `words[ix]` hoeft
dan niet meer, want het element komt al mee:

````python
for i, word in enumerate(words):
    print(i, word)
````

In een list comprehension werkt het net zo. Op welke posities staat `"spam"`?

````python
[i for i, word in enumerate(words) if word == "spam"]
````

`enumerate` begint bij 0 te tellen, net als een index. Wil je bij 1 beginnen,
bijvoorbeeld voor regelnummers, geef dan `1` mee als tweede argument:

````python
list(enumerate(["eerste", "tweede"], 1))
````

## Twee lijsten naast elkaar: `zip`

Soms horen twee lijsten bij elkaar, element voor element. Hier de namen van drie
spelers en hun scores:

````python
names = ["Anne", "Bram", "Cas"]
scores = [7, 9, 6]
````

`zip` legt ze naast elkaar, zoals een rits: het eerste element van de ene lijst
bij het eerste van de andere, het tweede bij het tweede, enzovoort. Ook dat
levert paren op:

````python
list(zip(names, scores))
````

Die paren pak je weer uit, in een lus of in een list comprehension:

````python
for name, score in zip(names, scores):
    print(name, score)
````

````python
[[name, score] for name, score in zip(names, scores)]
````

Zonder `zip` had je `range(len(names))` nodig en twee keer indexeren:
`[[names[i], scores[i]] for i in range(len(names))]`. Met `zip` zie je meteen
wat bij elkaar hoort. Zijn de lijsten niet even lang, dan stopt `zip` bij de
kortste.

## Per element een keuze: `a if c else b`

Soms wil je voor elk element een van twee waarden in de nieuwe lijst, afhankelijk
van een conditie. Met een `if`-statement schrijf je dat zo:

````python
temperature = -3

if temperature > 0:
    label = "dooi"
else:
    label = "vorst"

label
````

Python heeft er ook een uitdrukking voor, de **conditionele expressie**:

````python
label = "dooi" if temperature > 0 else "vorst"
label
````

Lees hem als: `"dooi"` als `temperature > 0`, en anders `"vorst"`. Het deel met
`else` is verplicht, want een uitdrukking levert altijd een waarde op.

Omdat het een uitdrukking is, past hij op de plek van *wat erin komt*:

````python
temperatures = [-3, 4, 0, 12]
["dooi" if t > 0 else "vorst" for t in temperatures]
````

Er kan nu op twee plekken een `if` staan, en die doen iets anders:

| | Voorbeeld | Wat het doet | Nieuwe lijst |
|---|---|---|---|
| filter, achteraan | `[t for t in temperatures if t > 0]` | laat elementen weg | `[4, 12]`, korter |
| conditionele expressie, vooraan | `["dooi" if t > 0 else "vorst" for t in temperatures]` | kiest per element een waarde | even lang als `temperatures` |

Een filter heeft geen `else`, een conditionele expressie wel.
