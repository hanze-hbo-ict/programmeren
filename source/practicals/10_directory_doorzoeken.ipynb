# Een directory doorzoeken

## Wat staat er allemaal in?

Een directory met subdirectories is een boom, en je weet van tevoren niet hoe
diep hij is. In het [eerste college](/lectures/10a_recursie) drukte je daarom met
recursie de namen af van alle bestanden in een directory. In dit werkcollege
doorzoek je een directory op meer manieren: hoeveel bestanden staan erin, hoeveel
ruimte nemen ze in, en staat er ergens een bestand dat er niet hoort? Een
mailprogramma weigert bijvoorbeeld een bijlage die op `.exe` eindigt, omdat zo'n
programma malware kan zijn.

### De directory

Een directory is een dictionary met drie sleutels, net als in het eerste college:

| Sleutel | Waarde |
|---|---|
| `"naam"` | de naam van de directory |
| `"bestanden"` | een dictionary van bestandsnaam naar grootte in kilobyte |
| `"directories"` | een lijst van subdirectories, elk weer een dictionary in deze vorm |

In de cel hieronder staat de directory `documents`, met drie subdirectories.
`projecten` heeft zelf ook weer een subdirectory:

```{mermaid}
flowchart TD
    D["documenten"] --> V(["verslag.docx"])
    D --> R(["rooster.pdf"])
    D --> P["projecten"]
    D --> F["foto's"]
    D --> L["leeg"]
    P --> PL(["plan.docx"])
    P --> LO(["logo.png"])
    P --> A["archief"]
    A --> N(["notulen.docx"])
    A --> S(["setup.exe"])
    F --> ST(["strand.jpg"])
    F --> B(["berg.jpg"])
```

`EMPTY` is een directory zonder bestanden en zonder subdirectories, en `PROJECTS`
is de subdirectory `projecten`. Voer de cel eerst uit.

````python
documents = {
    "naam": "documenten",
    "bestanden": {"verslag.docx": 120, "rooster.pdf": 300},
    "directories": [
        {
            "naam": "projecten",
            "bestanden": {"plan.docx": 40, "logo.png": 500},
            "directories": [
                {
                    "naam": "archief",
                    "bestanden": {"notulen.docx": 25, "setup.exe": 900},
                    "directories": [],
                },
            ],
        },
        {
            "naam": "foto's",
            "bestanden": {"strand.jpg": 2000, "berg.jpg": 1800},
            "directories": [],
        },
        {"naam": "leeg", "bestanden": {}, "directories": []},
    ],
}

EMPTY = {"naam": "leeg", "bestanden": {}, "directories": []}
PROJECTS = documents["directories"][0]
````

### Wat je gaat maken

| Stap | Functie | Doet | Hoe de recursie loopt |
|---|---|---|---|
| 1 | `count_files` | het aantal bestanden | een lus over de subdirectories |
| 2 | `total_size` | de totale grootte | `sum([...])` over de subdirectories |
| 3 | `has_extension` | eindigt een bestandsnaam op een extensie? | over een string, vanaf het eind |
| 4 | `find_files` | alle bestanden met een extensie | een lus over de subdirectories, met stap 3 |
| 5 | `depth` | hoeveel niveaus diep de directory is | een lus over de subdirectories |

Elke functie telt de directory zelf mee én al zijn subdirectories, hoe diep ze
ook gaan.

## Stap 1: `count_files(d)`

Geeft het aantal bestanden in `d` en in al zijn subdirectories.

| Aanroep | Resultaat |
|---|---|
| `count_files(documents)` | `8` |
| `count_files(PROJECTS)` | `4` |
| `count_files(EMPTY)` | `0` |

:::{admonition} Hint
:class: tip

Begin met het aantal bestanden in `d` zelf: `len(d["bestanden"])`. Tel daar in
een lus over `d["directories"]` voor elke subdirectory `count_files` van die
subdirectory bij op. Het basisgeval is een directory zonder subdirectories: dan
doet de lus niets.
:::

````python
# jouw oplossing
````

````python
assert count_files(documents) == 8
assert count_files(PROJECTS) == 4
assert count_files(EMPTY) == 0
````

## Stap 2: `total_size(d)`

Geeft de totale grootte in kilobyte van alle bestanden in `d` en in al zijn
subdirectories.

| Aanroep | Resultaat |

| `total_size(documents)` | `5685` |

| `total_size(EMPTY)` | `0` |

Schrijf het deze keer zonder lus, met `sum`.

:::{admonition} Hint

`d["bestanden"].values()` geeft de groottes van de bestanden in `d`, en `sum`
telt ze op. Voor de subdirectories maak je een list comprehension met voor elke
subdirectory zijn `total_size`, en die tel je ook op met `sum([...])`, zoals in
week 2.
:::

````python
# jouw oplossing
````

````python
assert total_size(documents) == 5685
assert total_size(PROJECTS) == 1465
assert total_size(EMPTY) == 0
````

## Stap 3: `has_extension(name, ext)`

Geeft `True` als de bestandsnaam `name` eindigt op de extensie `ext`, en anders
`False`. Gebruik recursie en vergelijk de strings vanaf het eind, teken voor
teken:

- Is `ext` leeg, dan geef je `True` terug: alle tekens van de extensie kwamen
  overeen.
- Is `name` leeg, dan geef je `False` terug: de naam is op, maar de extensie nog
  niet.
- Is het laatste teken van `name` niet gelijk aan het laatste teken van `ext`,
  dan geef je `False` terug.
- Anders roep je de functie opnieuw aan, met beide strings zonder hun laatste
  teken, en geef je het resultaat daarvan terug.

| Aanroep | Resultaat |
|---|---|
| `has_extension("setup.exe", ".exe")` | `True` |
| `has_extension("verslag.docx", ".exe")` | `False` |
| `has_extension("exe", ".exe")` | `False` |

:::{admonition} Hint
:class: tip

Het laatste teken van `s` is `s[-1]`, en alles ervóór is `s[:-1]`. Houd de
volgorde van de regels aan: zet de controle op een lege `ext` vóór die op een
lege `name`.
:::

````python
# jouw oplossing
````

````python
assert has_extension("setup.exe", ".exe") == True
assert has_extension("verslag.docx", ".exe") == False
assert has_extension("exe", ".exe") == False
assert has_extension("", "") == True
````

## Stap 4: `find_files(d, ext)`

Geeft een lijst met de namen van alle bestanden in `d` en in al zijn
subdirectories die eindigen op `ext`. Eerst de bestanden van `d` zelf, dan die
van de subdirectories, in de volgorde van de lijst.

| Aanroep | Resultaat |
|---|---|
| `find_files(documents, ".docx")` | `["verslag.docx", "plan.docx", "notulen.docx"]` |
| `find_files(documents, ".exe")` | `["setup.exe"]` |
| `find_files(documents, ".py")` | `[]` |

:::{admonition} Hint
:class: tip

De bestanden van `d` zelf vind je met een list comprehension en
`has_extension` uit stap 3. Loop daarna de subdirectories langs, en plak het
resultaat van `find_files` voor elke subdirectory achter je lijst met `+`.
:::

````python
# jouw oplossing
````

````python
assert find_files(documents, ".docx") == ["verslag.docx", "plan.docx", "notulen.docx"]
assert find_files(documents, ".exe") == ["setup.exe"]
assert find_files(documents, ".py") == []
````

## Stap 5: `depth(d)`

Geeft hoeveel niveaus diep `d` is. Een directory zonder subdirectories is 1 niveau
diep. Een directory met subdirectories is 1 niveau dieper dan zijn diepste
subdirectory.

| Aanroep | Resultaat |
|---|---|
| `depth(documents)` | `3` |
| `depth(PROJECTS)` | `2` |
| `depth(EMPTY)` | `1` |

:::{admonition} Hint
:class: tip

Onthoud in een lus over de subdirectories de grootste diepte tot nu toe, zoals je
in week 2 het langste woord tot nu toe onthield. Begin bij 0, zodat een directory
zonder subdirectories op `1 + 0` uitkomt.
:::

````python
# jouw oplossing
````

````python
assert depth(documents) == 3
assert depth(PROJECTS) == 2
assert depth(EMPTY) == 1
````
