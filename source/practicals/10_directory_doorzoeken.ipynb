# Een directory doorzoeken

## Wat staat er allemaal in?

Een directory met subdirectories is een boom, en je weet van tevoren niet hoe
diep hij is. In het [eerste college](/lectures/10a_recursie) drukte je daarom met
recursie de namen af van alle bestanden in een directory. In dit werkcollege
doorzoek je een directory op meer manieren: hoeveel bestanden staan erin, hoeveel
ruimte nemen ze in, en staat er ergens een bestand dat er niet hoort? Een
mailprogramma weigert bijvoorbeeld een bijlage die op `.exe` eindigt, omdat zo'n
programma malware kan zijn.

### De directory

Een directory is een dictionary met drie sleutels, net als in het eerste college:

| Sleutel | Waarde |
|---|---|
| `"naam"` | de naam van de directory |
| `"bestanden"` | een dictionary van bestandsnaam naar grootte in kilobyte |
| `"directories"` | een lijst van subdirectories, elk weer een dictionary in deze vorm |

In de cel hieronder staat de directory `documents`, met drie subdirectories.
`projecten` heeft zelf ook weer een subdirectory:

```{mermaid}
flowchart TD
    D["documenten"] --> V(["verslag.docx"])
    D --> R(["rooster.pdf"])
    D --> P["projecten"]
    D --> F["foto's"]
    D --> L["leeg"]
    P --> PL(["plan.docx"])
    P --> LO(["logo.png"])
    P --> A["archief"]
    A --> N(["notulen.docx"])
    A --> S(["setup.exe"])
    F --> ST(["strand.jpg"])
    F --> B(["berg.jpg"])
```

`EMPTY` is een directory zonder bestanden en zonder subdirectories, en `PROJECTS`
is de subdirectory `projecten`. Voer de cel eerst uit.

````python
documents = {
    "naam": "documenten",
    "bestanden": {"verslag.docx": 120, "rooster.pdf": 300},
    "directories": [
        {
            "naam": "projecten",
            "bestanden": {"plan.docx": 40, "logo.png": 500},
            "directories": [
                {
                    "naam": "archief",
                    "bestanden": {"notulen.docx": 25, "setup.exe": 900},
                    "directories": [],
                },
            ],
        },
        {
            "naam": "foto's",
            "bestanden": {"strand.jpg": 2000, "berg.jpg": 1800},
            "directories": [],
        },
        {"naam": "leeg", "bestanden": {}, "directories": []},
    ],
}

EMPTY = {"naam": "leeg", "bestanden": {}, "directories": []}
PROJECTS = documents["directories"][0]
````

### Wat je gaat maken

| Stap | Functie | Doet | Hoe de recursie loopt |
|---|---|---|---|
| 1 | `count_files` | het aantal bestanden | een lus over de subdirectories |
| 2 | `total_size` | de totale grootte | `sum([...])` over de subdirectories |
| 3 | `same_extension` | hebben twee bestandsnamen dezelfde extensie? | over twee strings, vanaf het eind |
| 4 | `find_similar` | alle bestanden met dezelfde extensie als een gegeven naam | een lus over de subdirectories, met stap 3 |
| 5 | `depth` | hoeveel niveaus diep de directory is | een lus over de subdirectories |

Elke functie telt de directory zelf mee én al zijn subdirectories, hoe diep ze
ook gaan.

## Stap 1: `count_files(d)`

Geeft het aantal bestanden in `d` en in al zijn subdirectories.

| Aanroep | Resultaat |
|---|---|
| `count_files(documents)` | `8` |
| `count_files(PROJECTS)` | `4` |
| `count_files(EMPTY)` | `0` |

:::{admonition} Hint
:class: tip

Begin met het aantal bestanden in `d` zelf: `len(d["bestanden"])`. Tel daar in
een lus over `d["directories"]` voor elke subdirectory `count_files` van die
subdirectory bij op. Het basisgeval is een directory zonder subdirectories: dan
doet de lus niets.
:::

````python
# jouw oplossing
````

````python
assert count_files(documents) == 8
assert count_files(PROJECTS) == 4
assert count_files(EMPTY) == 0
````

| Aanroep | Resultaat |

Schrijf het deze keer zonder lus, met `sum`.

`d["bestanden"].values()` geeft de groottes van de bestanden in `d`, en `sum`

:::

````python
# jouw oplossing
````

````python
assert total_size(documents) == 5685
assert total_size(PROJECTS) == 1465
assert total_size(EMPTY) == 0
````

## Stap 3: `same_extension(name1, name2)`

De **extensie** van een bestandsnaam is het stuk vanaf de laatste punt:
`verslag.docx` heeft de extensie `.docx`. Een naam zonder punt heeft geen
extensie.

`same_extension(name1, name2)` geeft `True` als de bestandsnamen `name1` en
`name2` dezelfde extensie hebben, en anders `False`. Gebruik recursie en
vergelijk de twee namen vanaf het eind, teken voor teken:

- Is een van de twee namen leeg, dan geef je `False` terug: die naam had geen
  punt, en dus geen extensie.
- Is het laatste teken van `name1` niet gelijk aan het laatste teken van
  `name2`, dan geef je `False` terug.
- Is het laatste teken een punt, dan geef je `True` terug: alles vanaf de punt
  was gelijk.
- Anders roep je de functie opnieuw aan, met beide namen zonder hun laatste
  teken, en geef je het resultaat daarvan terug.

| Aanroep | Resultaat |
|---|---|
| `same_extension("plan.docx", "verslag.docx")` | `True` |
| `same_extension("logo.png", "strand.jpg")` | `False` |
| `same_extension("readme", "readme")` | `False` |

:::{admonition} Hint
:class: tip

Het laatste teken van `s` is `s[-1]`, en alles ervóór is `s[:-1]`. Elke regel
wordt één tak van je `if`, in dezelfde volgorde. Bij de derde regel weet je al dat
de twee laatste tekens gelijk zijn, dus het is genoeg om er één te bekijken.
:::

````python
# jouw oplossing
````

````python
assert same_extension("plan.docx", "verslag.docx") == True
assert same_extension("x.py", "y.py") == True
assert same_extension("logo.png", "strand.jpg") == False
assert same_extension("setup.exe", "notulen.docx") == False
assert same_extension("readme", "readme") == False
````

## Stap 4: `find_similar(d, name)`

Geeft een lijst met de namen van alle bestanden in `d` en in al zijn
subdirectories die dezelfde extensie hebben als `name`. Eerst de bestanden van
`d` zelf, dan die van de subdirectories, in de volgorde van de lijst. `name` zelf
hoeft niet in `d` te staan: met `find_similar(documents, "virus.exe")` vind je
alle `.exe`-bestanden.

| Aanroep | Resultaat |
|---|---|
| `find_similar(documents, "brief.docx")` | `["verslag.docx", "plan.docx", "notulen.docx"]` |
| `find_similar(documents, "virus.exe")` | `["setup.exe"]` |
| `find_similar(documents, "script.py")` | `[]` |

:::{admonition} Hint
:class: tip

De bestanden van `d` zelf vind je met een list comprehension en
`same_extension` uit stap 3. Loop daarna de subdirectories langs, en plak het
resultaat van `find_similar` voor elke subdirectory achter je lijst met `+`.
:::

````python
# jouw oplossing
````

````python
assert find_similar(documents, "brief.docx") == [
    "verslag.docx",
    "plan.docx",
    "notulen.docx",
]
assert find_similar(documents, "virus.exe") == ["setup.exe"]
assert find_similar(documents, "script.py") == []
````

## Stap 5: `depth(d)`

Geeft hoeveel niveaus diep `d` is. Een directory zonder subdirectories is 1 niveau
diep. Een directory met subdirectories is 1 niveau dieper dan zijn diepste
subdirectory.

| Aanroep | Resultaat |
|---|---|
| `depth(documents)` | `3` |
| `depth(PROJECTS)` | `2` |
| `depth(EMPTY)` | `1` |

:::{admonition} Hint
:class: tip

Onthoud in een lus over de subdirectories de grootste diepte tot nu toe, zoals je
in week 2 het langste woord tot nu toe onthield. Begin bij 0, zodat een directory
zonder subdirectories op `1 + 0` uitkomt.
:::

````python
# jouw oplossing
````

````python
assert depth(documents) == 3
assert depth(PROJECTS) == 2
assert depth(EMPTY) == 1
````
