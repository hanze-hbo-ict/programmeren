## Opdracht 1

`a.switch()` verandert alleen het attribuut `on` van `a`. De uitvoer is `True` en daarna `False`.

```python
class Lamp:
    """Een lamp die aan of uit staat."""

    def __init__(self, color):
        """Maak een lamp met de gegeven kleur; hij staat uit."""
        self.color = color
        self.on = False

    def switch(self):
        """Zet de lamp aan als hij uit staat, en uit als hij aan staat."""
        self.on = not self.on


a = Lamp("rood")
b = Lamp("blauw")
a.switch()
print(a.on)
print(b.on)
```

object `a`; `b` blijft uit.

## Opdracht 3

De klasse `Point` hieronder bevat ook de methoden van opdracht 4, 5 en 6.

```python
class Point:
    """Een punt in het vlak."""

    def __init__(self, x, y):
        """Maak een punt met de gegeven coördinaten."""
        self.x = x
        self.y = y

    def __repr__(self):
        """Geeft het punt als string, zoals Point(3, 4)."""
        return f"Point({self.x}, {self.y})"

    def distance_to_origin(self):
        """Geeft de afstand tot het punt (0, 0)."""
        return (self.x * self.x + self.y * self.y) ** 0.5

    def move(self, dx, dy):
        """Verschuift het punt over dx en dy."""
        self.x += dx
        self.y += dy
```

```python
p = Point(3, 4)
assert p.x == 3
assert p.y == 4
```

## Opdracht 5

```python
assert Point(3, 4).distance_to_origin() == 5.0
assert Point(0, 0).distance_to_origin() == 0.0
```

## Opdracht 8

De klasse `Square` hieronder bevat ook de methode `grow` van opdracht 9. `area` is een property, dus je schrijft `s.area` zonder haakjes.

```python
class Square:
    """Een vierkant waarvan de zijde alleen groter kan worden."""

    def __init__(self, side):
        """Maak een vierkant met de gegeven zijde."""
        self._side = side

    @property
    def side(self):
        """De lengte van de zijde, alleen om te lezen."""
        return self._side

    @property
    def area(self):
        """De oppervlakte."""
        return self._side * self._side

    def grow(self, amount):
        """Maakt de zijde amount langer, maar alleen als amount positief is."""
        if amount > 0:
            self._side += amount
```

```python
assert Square(3).area == 9
assert Square(0).area == 0
```

## Opdracht 9

`grow` staat in de klasse en mag `self._side` dus veranderen, maar alleen als `amount` positief is.

```python
s = Square(3)
s.grow(2)
assert s.side == 5
s.grow(-4)
assert s.side == 5
s.grow(0)
assert s.side == 5
```

## Opdracht 10

Van buiten de klasse kan de zijde alleen via `grow` veranderen, en `grow` maakt
haar alleen groter. `s.side = 1` wordt geweigerd, want `side` heeft geen setter.

Met `s._side = 1` kan het wel: Python houdt dat niet tegen. De `_` is een
afspraak dat je zo'n attribuut van buiten de klasse niet gebruikt.

```python
s = Square(3)
s._side = 1
print(s.side)
```

## Opdracht 11

`Figure` bewaart de punten in `self._points` en laat ze zelf verschuiven: `move`
roept voor elk punt zijn eigen `move` aan. De klasse `Point` is die van opdracht 3.

```python
class Figure:
    """Een figuur die uit punten bestaat."""

    def __init__(self):
        """Maak een lege figuur."""
        self._points = []

    def add(self, point):
        """Voeg point toe aan de figuur."""
        self._points.append(point)

    def move(self, dx, dy):
        """Verschuift elk punt van de figuur over dx en dy."""
        for point in self._points:
            point.move(dx, dy)

    def __repr__(self):
        """Geeft de figuur als string, zoals Figure([Point(0, 0)])."""
        return f"Figure({self._points})"
```

```python
p = Point(0, 0)
q = Point(3, 4)
f = Figure()
f.add(p)
f.add(q)
assert repr(f) == "Figure([Point(0, 0), Point(3, 4)])"
f.move(1, 1)
assert repr(f) == "Figure([Point(1, 1), Point(4, 5)])"
assert repr(p) == "Point(1, 1)"
p.move(-1, -1)
assert repr(f) == "Figure([Point(0, 0), Point(4, 5)])"
empty = Figure()
empty.move(1, 1)
assert repr(empty) == "Figure([])"
```

`f.add(p)` zet geen kopie van `p` in de figuur, maar `p` zelf. `f.move(1, 1)`
verschuift dus het punt `p`, en `p.move(-1, -1)` verschuift het punt dat in de
figuur zit. Bij een lege figuur loopt de lus in `move` nul keer.
