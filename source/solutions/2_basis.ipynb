## Opdracht 1a

De klant betaalt €60 startkosten en €40 per uur. Eerst lezen we het aantal uren als integer. Daarna berekenen we de rekening met `60 + 40 * uren` en drukken we die af.


```python
uren = int(input("Aantal uren: "))
rekening = 60 + 40 * uren
print(rekening)
```

## Opdracht 1b

De basisrekening berekenen we zoals bij 1a. Is die hoger dan €400, dan vermenigvuldigen we haar met `0.9`; dat is de rekening na 10 procent korting. Daarna drukken we het resultaat af.


```python
uren = int(input("Aantal uren: "))
rekening = 60 + 40 * uren
if rekening > 400:
    rekening = rekening * 0.9
print(rekening)
```

## Opdracht 2

Het cijfer volgt uit `punten / max_punten * 9 + 1`. We lezen beide waarden als integers, voeren de berekening stap voor stap uit in `cijfer` en drukken het resultaat af.

## Opdracht 3a

De gebruiker geeft een temperatuur en een schaal. Bij Fahrenheit trekken we eerst 32 af en vermenigvuldigen we daarna met `5 / 9`. In het andere geval rekenen we Celsius om met `temperatuur * 9 / 5 + 32`. De `if` kiest de juiste formule.


```python
temperatuur = float(input("Temperatuur?: "))
schaal = input("Schaal?: ")
if schaal == "Fahrenheit":
    celsius = (temperatuur - 32) * 5 / 9
    print("Celsius =", celsius)
else:
    fahrenheit = temperatuur * 9 / 5 + 32
    print("Fahrenheit =", fahrenheit)
```

## Opdracht 3b

Naast de twee schalen uit 3a behandelen we Kelvin. Bij Kelvin trekken we `273.15` af om Celsius te krijgen en gebruiken we daarna de formule uit 3a voor Fahrenheit. Bij Celsius berekenen we Fahrenheit en Kelvin; bij Fahrenheit berekenen we Celsius en daarna Kelvin.


```python
temperatuur = float(input("Temperatuur?: "))
schaal = input("Schaal?: ")
if schaal == "Kelvin":
    celsius = temperatuur - 273.15
    fahrenheit = celsius * 9 / 5 + 32
    print("Celsius =", celsius)
    print("Fahrenheit =", fahrenheit)
elif schaal == "Celsius":
    fahrenheit = temperatuur * 9 / 5 + 32
    kelvin = temperatuur + 273.15
    print("Fahrenheit =", fahrenheit)
    print("Kelvin =", kelvin)
else:
    celsius = (temperatuur - 32) * 5 / 9
    kelvin = celsius + 273.15
    print("Celsius =", celsius)
    print("Kelvin =", kelvin)
```
