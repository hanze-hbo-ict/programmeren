# Extra: van tekst naar beeld

Deze uitwerking gebruikt `patroon.txt` als tekstdata. `png.py` hoort bij de bestaande beeldroute uit `fraaie_plaatjes.zip`; pak die zip uit voordat je de laatste functie uitvoert.

```python
def lees_raster(bestandsnaam):
    """Leest een niet-leeg rechthoekig raster met . en #."""
    with open(bestandsnaam) as bestand:
        raster = []
        for regel in bestand:
            raster.append(regel.rstrip("\n"))
    assert len(raster) > 0
    breedte = len(raster[0])
    assert breedte > 0
    for regel in raster:
        assert len(regel) == breedte
        for teken in regel:
            assert teken == "." or teken == "#"
    return raster
```

```python
def raster_naar_pixels(raster, schaal):
    """Maakt RGB-pixels met iedere rastercel als vierkant blok."""
    assert len(raster) > 0
    assert schaal == 8
    breedte = len(raster[0])
    assert breedte > 0
    for regel in raster:
        assert len(regel) == breedte
    pixels = []
    for regel in raster:
        uitgerekte_regel = []
        for teken in regel:
            if teken == ".":
                pixel = [255, 255, 255]
            else:
                assert teken == "#"
                pixel = [0, 0, 0]
            for kolom in range(schaal):
                uitgerekte_regel.append(pixel)
        for rij in range(schaal):
            pixels.append(uitgerekte_regel)
    return pixels
```

```python
try:
    from png import save_rgb
    png_beschikbaar = True
except ImportError:
    png_beschikbaar = False

def schrijf_patronenbeeld(bestandsnaam, raster):
    """Schrijft raster als patroon.png via de bestaande png-module."""
    pixels = raster_naar_pixels(raster, 8)
    assert len(pixels) == len(raster) * 8
    assert len(pixels[0]) == len(raster[0]) * 8
    assert png_beschikbaar
    save_rgb(pixels, bestandsnaam)
    return bestandsnaam
```

```python
raster = lees_raster("../problems/assets/patroon.txt")
assert len(raster) == 8
assert len(raster[0]) == 12
pixels = raster_naar_pixels(raster, 8)
assert len(pixels) == 64
assert len(pixels[0]) == 96
assert pixels[0][0] == [255, 255, 255]
assert pixels[0][2 * 8] == [0, 0, 0]
assert pixels[4 * 8][2 * 8] == [0, 0, 0]
klein = raster_naar_pixels([".#"], 8)
assert len(klein) == 8
assert len(klein[0]) == 16
assert klein[0][0] == [255, 255, 255]
assert klein[0][8] == [0, 0, 0]
assert schrijf_patronenbeeld.__name__ == "schrijf_patronenbeeld"
# Na het uitpakken van fraaie_plaatjes.zip: schrijf_patronenbeeld("patroon.png", raster)
```
