Tijdens deze opdracht ga je een programma schrijven om de punten in en rond de mandelbrotverzameling weer te geven en te verkennen.

De vier rekenfuncties hieronder staan in cellen die bij de build worden uitgevoerd, met hun assertions eronder. De functies die een PNG-bestand tekenen staan daaronder als codeblok: die hebben de module `png` uit `mandelbrot.zip` nodig en schrijven een bestand weg, dus ze horen thuis op je eigen machine en niet in de build.

### De functie `mult`

````python
def mult(c, n):
    """Vermenigvuldigt c met de positieve integer n, met alleen een lus en optellen."""
    result = 0
    for i in range(n):
        result += c
    return result


assert mult(3, 5) == 15
assert mult(6, 7) == 42
assert mult(1.5, 28) == 42.0
````

### De functie `update`

````python
def update(c, n):
    """Begint met z = 0, voert z = z**2 + c n keer uit en geeft de laatste z terug."""
    z = 0
    for i in range(n):
        z = z**2 + c
    return z


assert update(1, 3) == 5
assert update(-1, 3) == -1
assert update(-1, 10) == 0
````

### De functie `in_mset`

````python
def in_mset(c, n):
    """Bepaalt of c binnen de mandelbrotverzameling lijkt te vallen.

    c is de waarde uit de stap z = z**2 + c, en n het aantal keren dat
    die stap hoogstens wordt uitgevoerd.

    Geeft False terug zodra abs(z) groter wordt dan 2, en True als dat
    binnen n stappen niet gebeurt.
    """
    z = 0
    for i in range(n):
        z = z**2 + c
        if abs(z) > 2:
            return False
    return True


c = 0 + 0j
assert in_mset(c, 25) == True
c = 3 + 4j
assert in_mset(c, 25) == False
c = 0.3 + -0.5j
assert in_mset(c, 25) == True
c = -0.7 + 0.3j
assert in_mset(c, 25) == False
c = 0.42 + 0.2j
assert in_mset(c, 25) == True
assert in_mset(c, 50) == False
````

### De functie `scale`

````python
def scale(pix, pix_max, float_min, float_max):
    """Geeft de floating-pointwaarde terug die bij pixel pix hoort.

    pix is de huidige pixelkolom (of pixelrij), pix_max het totale aantal
    pixelkolommen, en float_min en float_max zijn de onder- en bovengrens
    van het bereik.
    """
    return float_min + (pix / pix_max * (float_max - float_min))


assert scale(100, 200, -2.0, 1.0) == -0.5
assert scale(100, 200, -1.5, 1.5) == 0.0
assert scale(100, 300, -2.0, 1.0) == -1.0
assert scale(25, 300, -2.0, 1.0) == -1.75
````
