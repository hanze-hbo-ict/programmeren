## Opdracht 10

```python
kleuren = ["rood", "blauw", "rood", "groen", "rood", "blauw"]

aantallen = {}
for kleur in kleuren:
    if kleur not in aantallen:
        aantallen[kleur] = 1
    else:
        aantallen[kleur] += 1

print(aantallen)
```

## Opdracht 11

```python
L = [3, 1, 4]
nieuw = L + [9]

M = [3, 1, 4]
M.append(9)

assert nieuw == M
print(nieuw)
print(M)
```

## Opdracht 12

De volgorde van de namen in een set ligt niet vast. Die kan bij jou anders
zijn.

```python
gasten = ["Anne", "Bram", "Anne", "Cas", "Bram", "Anne"]
unieke_gasten = set(gasten)
print(unieke_gasten)
```

## Opdracht 13

```python
print("Cas" in unieke_gasten)
print("Daan" in unieke_gasten)
```

## Opdracht 14

```python
unieke_gasten.add("Daan")

assert "Daan" in unieke_gasten
```
