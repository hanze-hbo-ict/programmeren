## Opdracht 1

`sorted` en `max` zonder `key=` vergelijken de woorden op alfabet: `"zebra"` is
dan het grootste. Met `key=len` vergelijken ze de lengte, en dan is `"olifant"`
het grootste en `"kat"` het kleinste.

````python
animals = ["olifant", "kat", "zebra"]
print(sorted(animals))
print(max(animals))
print(max(animals, key=len))
print(min(animals, key=len))
````

## Opdracht 2

Zonder `key=` sorteert `sorted` op de hele string, dus op de letter vooraan. Met
de lambda-functie sorteert ze op het tweede teken, het cijfer. Voor `"b2"` geeft
de lambda-functie `"2"` terug.

````python
codes = ["b2", "a3", "c1"]
print(sorted(codes))
print(sorted(codes, key=lambda s: s[1]))
````

## Opdracht 3

`function` roept `f` twee keer aan: eerst op `x`, en dan op wat daaruit komt. Bij
`10` wordt dat `10 + 3 + 3`, en bij `"hoi"` komen er twee uitroeptekens achter.

````python
def function(f, x):
    return f(f(x))


print(function(lambda x: x + 3, 10))
print(function(lambda s: s + "!", "hoi"))
````

## Opdracht 4

Het programma drukt zeven aanroepen af. Elke aanroep die niet in een basisgeval
eindigt, doet er twee: eerst *use it*, dan *lose it*.

```text
function(3, [1, 2])          True
├── function(2, [2])         True    use it: de 1 gebruikt
│   ├── function(0, [])      True    use it: de 2 gebruikt
│   └── function(2, [])      False   lose it
└── function(3, [2])         False   lose it: de 1 laten liggen
    ├── function(1, [])      False   use it
    └── function(3, [])      False   lose it
```

De laatste regel drukt `True` af: 3 cent betaal je met 1 en 2.

````python
def function(amount, coins):
    print(amount, coins)
    if amount == 0:
        return True
    elif amount < 0:
        return False
    elif coins == []:
        return False
    else:
        use_it = function(amount - coins[0], coins[1:])
        lose_it = function(amount, coins[1:])
        return use_it or lose_it


print(function(3, [1, 2]))
````

## Opdracht 5

`function(5, [4, 3, 2])` geeft `5` terug, met de items 3 en 2. *Use it* met de 4
levert maar 4 op: daarna is er nog 1 over, en bij de aanroepen
`function(1, [3, 2])` en `function(1, [2])` past het eerste item niet. Daar
wordt alleen *lose it* geprobeerd.

````python
def function(capacity, items):
    if items == []:
        return 0
    first = items[0]
    rest = items[1:]
    if first > capacity:
        return function(capacity, rest)
    else:
        use_it = first + function(capacity - first, rest)
        lose_it = function(capacity, rest)
        return max(use_it, lose_it)


print(function(5, [4, 3, 2]))
````

## Opdracht 7

````python
pairs = [["b", 2], ["a", 3], ["c", 1]]


def second(pair):
    """Geeft het tweede element van pair."""
    return pair[1]


by_second = sorted(pairs, key=second)
by_second_lambda = sorted(pairs, key=lambda pair: pair[1])

assert second(["b", 2]) == 2
assert by_second == [["c", 1], ["b", 2], ["a", 3]]
assert by_second_lambda == [["c", 1], ["b", 2], ["a", 3]]
````

## Opdracht 9

Zodra één element niet aan de test voldoet, weet je het antwoord en stopt de
recursie.

````python
def all_pass(L, test):
    """Geeft True als test(x) waar is voor elk element x van L."""
    if L == []:  # basisgeval
        return True
    elif not test(L[0]):  # basisgeval: dit element voldoet niet
        return False
    else:  # recursief geval
        return all_pass(L[1:], test)


assert all_pass([2, 4, 6], lambda x: x % 2 == 0) == True
assert all_pass([2, 3, 4], lambda x: x % 2 == 0) == False
assert all_pass(["kat", "ezel"], lambda s: len(s) >= 3) == True
assert all_pass([], lambda x: x > 0) == True
````
