## Opdracht 1

`len(w)` komt in de lijst, niet `w` zelf. Het filter in regel 3 houdt de
veelvouden van 3 over, en 0 hoort daarbij. In regel 4 komen de posities van de
klinkers erin, niet de klinkers. In regel 6 kiest de conditionele expressie per
getal een woord; bij 5 is `x > 5` niet waar.

```python
print([len(w) for w in ["de", "vuurtoren", "draait"]])
print([w for w in ["de", "vuurtoren", "draait"] if len(w) > 2])
print([x for x in range(10) if x % 3 == 0])
print([i for i, c in enumerate("toren") if c in "aeiou"])
print([a + b for a, b in zip([1, 2, 3], [10, 20, 30])])
print(["hoog" if x > 5 else "laag" for x in [3, 8, 5]])
```

## Opdracht 2

```python
def squares(n):
    """Geeft een lijst met de kwadraten van 1 tot en met n."""
    return [x * x for x in range(1, n + 1)]


assert squares(0) == []
assert squares(1) == [1]
assert squares(3) == [1, 4, 9]
```

## Opdracht 3

```python
def without(L, x):
    """Geeft de elementen uit L die niet gelijk zijn aan x."""
    return [element for element in L if element != x]


assert without([1, 0, 2, 0], 0) == [1, 2]
assert without(["de", "toren"], "licht") == ["de", "toren"]
assert without([], 0) == []
```

## Opdracht 4

```python
def positions(L, x):
    """Geeft de posities waarop x in L staat."""
    return [i for i, element in enumerate(L) if element == x]


assert positions(["de", "toren", "de"], "de") == [0, 2]
assert positions(["de", "toren"], "licht") == []
assert positions([], "de") == []
```

## Opdracht 5

```python
def highest(L1, L2):
    """Geeft per positie het grootste van de twee getallen."""
    return [max(a, b) for a, b in zip(L1, L2)]


assert highest([1, 9, 5], [4, 2, 5]) == [4, 9, 5]
assert highest([-1, -9], [-4, -2]) == [-1, -2]
assert highest([], []) == []
```

## Opdracht 6

Het resultaat is even lang als `temperatures`: de conditionele expressie kiest per element een waarde en laat niets weg.

```python
def frost(temperatures):
    """Geeft per temperatuur "vorst" bij 0 graden of kouder, en anders "dooi"."""
    return ["vorst" if t <= 0 else "dooi" for t in temperatures]


assert frost([-1, 0, 10]) == ["vorst", "vorst", "dooi"]
assert frost([21, -21]) == ["dooi", "vorst"]
assert frost([]) == []
```

## Opdracht 7

Regel 1 is een set: elke letter staat er één keer in, ook de `u` en de `r`
die dubbel in het woord staan. Regel 3 is ook een set, met de drie mogelijke
resten bij deling door 3. Regel 4 is een lijst van lijsten met één `for`; regel 5
is een geneste comprehension, met twee rijen van drie.

```python
print({c for c in "vuurtoren"})
print({w: len(w) for w in ["licht", "nacht"]})
print({x % 3 for x in range(10)})
print([[y] * 2 for y in range(3)])
print([[x * y for x in range(1, 4)] for y in range(1, 3)])
```

## Opdracht 8

```python
def first_letters(words):
    """Geeft de set van beginletters van de woorden."""
    return {word[0] for word in words}


assert first_letters(["de", "vuurtoren", "draait"]) == {"d", "v"}
assert first_letters(["licht"]) == {"l"}
assert first_letters([]) == set()
```

## Opdracht 9

Een woord dat twee keer voorkomt, wordt twee keer als sleutel gezet, met dezelfde waarde. In de dictionary staat het één keer.

```python
def word_lengths(words):
    """Geeft een dictionary van elk woord naar zijn lengte."""
    return {word: len(word) for word in words}


assert word_lengths(["de", "toren"]) == {"de": 2, "toren": 5}
assert word_lengths(["de", "de"]) == {"de": 2}
assert word_lengths([]) == {}
```

## Opdracht 10

```python
def common(counts, n):
    """Geeft de woorden uit counts die minstens n keer voorkomen, met hun aantal."""
    return {word: count for word, count in counts.items() if count >= n}


assert common({"de": 3, "toren": 1, "licht": 2}, 2) == {"de": 3, "licht": 2}
assert common({"de": 3}, 4) == {}
assert common({}, 2) == {}
```

## Opdracht 11

```python
def count_above(numbers, limit):
    """Geeft het aantal getallen in numbers dat groter is dan limit."""
    return sum([1 for x in numbers if x > limit])


assert count_above([3, 8, 5], 4) == 2
assert count_above([3, 8, 5], 10) == 0
assert count_above([], 4) == 0
```

## Opdracht 12

Een geneste comprehension kan ook: `[["." for _ in range(width)] for _ in range(height)]`. Allebei goed.

```python
def grid(width, height):
    """Geeft een raster van height rijen met elk width punten."""
    return [["."] * width for _ in range(height)]


assert grid(3, 2) == [[".", ".", "."], [".", ".", "."]]
assert grid(1, 1) == [["."]]
assert grid(3, 0) == []
```

## Opdracht 13

```python
def staircase(n):
    """Geeft een trap van n rijen, waarin rij y de getallen 1 tot en met y bevat."""
    return [[x for x in range(1, y + 1)] for y in range(1, n + 1)]


assert staircase(0) == []
assert staircase(1) == [[1]]
assert staircase(3) == [[1], [1, 2], [1, 2, 3]]
```
