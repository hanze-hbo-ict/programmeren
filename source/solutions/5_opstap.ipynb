## Opdracht 1

De binnenste lus draait drie keer voor elke ronde van de buitenste lus: `harry` verschijnt negen keer.

```python
name = "harry"
for x in range(3):
    for x in range(3):
        print(name)
```

## Opdracht 2

De binnenste lus loopt drie keer langs `lst`, voor elk van de drie rondes van de buitenste lus.

```python
lst = [1, 2, 3]
for ix1 in range(3):
    for ix2 in lst:
        print(ix2)
```

## Opdracht 3

De uitvoer loopt eerst door de eerste rij en daarna door de tweede rij.

```python
nested_lst = [[1, 2, 3], [9, 8, 7]]
for ix1 in nested_lst:
    for ix2 in ix1:
        print(ix2)
```

## Opdracht 4

Deze versie gebruikt indices en heeft dezelfde uitvoer als opdracht 3: `1`, `2`, `3`, `9`, `8`, `7`.

```python
nested_lst = [[1, 2, 3], [9, 8, 7]]
for ix1 in range(len(nested_lst)):
    for ix2 in range(len(nested_lst[ix1])):
        print(nested_lst[ix1][ix2])
```

## Opdracht 5

De eerste `print` geeft `[[40, 50], [800, 900]]`. De tweede probeert rij 2 van `grid` te lezen; die bestaat niet, dus het programma eindigt met een `IndexError`.

```python
grid = [[4, 5], [8, 9], [1, 2]]
k = 1
new_grid = []
for i in range(2):
    k = k * 10
    new_row = []
    for j in range(2):
        new_row = new_row + [grid[i][j] * k]
    new_grid = new_grid + [new_row]
print(new_grid)
print(grid[2][0])
```

## Opdracht 7

De eerste regel is leeg; daarna verschijnen `*`, `**`, `***` en `****`.

```python
n = 5
for i in range(n):
    for j in range(i):
        print("*", end="")
    print("")
```

## Opdracht 8

Na de oplopende driehoek volgt een regel met vijf tekens, daarna vier, drie, twee en één.

```python
n = 5
for i in range(n):
    for j in range(i):
        print("*", end="")
    print("")
for i in range(n, 0, -1):
    for j in range(i):
        print("*", end="")
    print("")
```

## Opdracht 10

De binnenste lus bouwt per rij een `new_row`; de buitenste lus verzamelt die rijen in de 2D-lijst `grid`.

```python
grid = []
for r in range(3):
    new_row = []
    for k in range(4):
        new_row = new_row + [r * k]
    grid = grid + [new_row]

assert grid == [[0, 0, 0, 0], [0, 1, 2, 3], [0, 2, 4, 6]]
```

## Opdracht 11

Het aantal rijen en kolommen staat direct in de 2D-lijst; daarom is geen lus nodig.

```python
def size(grid):
    """Geeft het aantal rijen en het aantal kolommen van grid terug."""
    return len(grid), len(grid[0])


assert size([[1, 2, 3], [4, 5, 6]]) == (2, 3)
assert size([[7]]) == (1, 1)
```

## Opdracht 12

`column_values` is de verzamelvariabele. De lus loopt langs de rijen en voegt uit elke rij het element op kolom `k` toe.

```python
def column(grid, k):
    """Geeft kolom k van grid als lijst terug."""
    column_values = []
    for r in range(len(grid)):
        column_values = column_values + [grid[r][k]]
    return column_values


assert column([[1, 2, 3], [4, 5, 6]], 1) == [2, 5]
assert column([[1, 2, 3], [4, 5, 6]], 0) == [1, 4]
assert column([[7]], 0) == [7]
```
