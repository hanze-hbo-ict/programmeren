# Opstap

De uitwerkingen staan in dezelfde volgorde als de dertien opdrachten in `problems/13_opstap`.

De klasse `Lamp` waar alle opdrachten op voortbouwen:

```python
class Lamp:
    """Een lamp die aan of uit staat."""

    def __init__(self, color):
        """Maak een lamp met de gegeven kleur; hij staat uit."""
        self.color = color
        self.on = False

    def switch(self):
        """Zet de lamp aan als hij uit staat, en uit als hij aan staat."""
        self.on = not self.on

    def describe(self):
        """Geeft een korte beschrijving, zoals lamp rood: uit."""
        if self.on:
            return f"lamp {self.color}: aan"
        return f"lamp {self.color}: uit"
```

## Opdracht 1

`NightLamp` erft alles van `Lamp`. De uitvoer is `True` en daarna
`lamp geel: aan`.

```python
class NightLamp(Lamp):
    """Een nachtlampje."""


n = NightLamp("geel")
n.switch()
print(n.on)
print(n.describe())
```

## Opdracht 2

Uit de superklasse `Lamp`. Een subklasse krijgt alle methoden van haar
superklasse mee. `NightLamp("geel")` roept dus de constructor van `Lamp` aan, en
`n.switch()` en `n.describe()` draaien de versies uit `Lamp`.

## Opdracht 3

De klasse hieronder bevat ook de standaardwaarde van opdracht 4 en de methode
`describe` van opdracht 5. `super().__init__(color)` geeft `color` en `on` hun
beginwaarde; daarna zet de constructor `level`.

```python
class DimmableLamp(Lamp):
    """Een lamp die je kunt dimmen."""

    def __init__(self, color, level=100):
        """Maak een dimbare lamp; zonder level brandt hij op 100%."""
        super().__init__(color)
        self.level = level

    def describe(self):
        """Geeft de beschrijving van een lamp, met de helderheid erachter."""
        return super().describe() + f", {self.level}%"
```

```python
d = DimmableLamp("wit", 30)
assert d.color == "wit"
assert not d.on
assert d.level == 30
d.switch()
assert d.on
```

## Opdracht 4

Met `level=100` in de signatuur mag je het argument weglaten. Geef je het wel mee,
dan geldt wat je meegeeft, ook als dat `0` is.

```python
assert DimmableLamp("wit").level == 100
assert DimmableLamp("wit", 30).level == 30
assert DimmableLamp("wit", 0).level == 0
```

## Opdracht 5

`super().describe()` geeft de tekst van `Lamp`, zoals `lamp wit: uit`. De
methode van `DimmableLamp` plakt daar alleen de helderheid achter.

```python
d = DimmableLamp("wit", 30)
assert d.describe() == "lamp wit: uit, 30%"
d.switch()
assert d.describe() == "lamp wit: aan, 30%"
assert DimmableLamp("blauw").describe() == "lamp blauw: uit, 100%"
```

## Opdracht 6

De uitvoer is `50`, en daarna een foutmelding:
`AttributeError: 'BrokenLamp' object has no attribute 'on'`.

```python
class BrokenLamp(Lamp):
    """Een lamp met een constructor die iets vergeet."""

    def __init__(self, color, level):
        """Maak een lamp met de gegeven kleur en helderheid."""
        self.level = level


b = BrokenLamp("rood", 50)
print(b.level)
print(b.describe())
```

De attributen `on` en `color` ontbreken; de foutmelding noemt alleen `on`, omdat
`describe` daar het eerst op vastloopt. Die twee krijgen hun beginwaarde in
de constructor van `Lamp`, en die wordt nu niet aangeroepen: de constructor van
`BrokenLamp` overschrijft hem. `describe` van `Lamp` leest `self.on` en loopt
daarop vast. In de constructor hoort als eerste regel
`super().__init__(color)`.

## Opdracht 7

`switch` roept eerst de versie van `Lamp` aan, en telt daarna.

```python
class CountingLamp(Lamp):
    """Een lamp die telt hoe vaak hij is omgezet."""

    def __init__(self, color):
        """Maak een lamp die nog nooit is omgezet."""
        super().__init__(color)
        self.count = 0

    def switch(self):
        """Zet de lamp om, en telt dat."""
        super().switch()
        self.count += 1
```

```python
c = CountingLamp("blauw")
assert c.count == 0
assert not c.on
c.switch()
c.switch()
c.switch()
assert c.count == 3
assert c.on
assert c.describe() == "lamp blauw: aan"
```

## Opdracht 8

De uitvoer is `lamp groen: aan, 60%`. `ReadingLamp` heeft zelf geen `describe`,
dus draait die van haar superklasse `DimmableLamp`. Die roept op haar beurt met
`super().describe()` de versie van `Lamp` aan, en die zet `lamp groen: aan` in de
tekst.

```python
class ReadingLamp(DimmableLamp):
    """Een leeslamp: een dimbare lamp die op 60% staat."""

    def __init__(self, color):
        """Maak een leeslamp op 60%."""
        super().__init__(color, 60)


r = ReadingLamp("groen")
r.switch()
print(r.describe())
```

## Opdracht 9

Elke lamp gebruikt haar eigen `describe`, of die van de dichtstbijzijnde
superklasse die er een heeft:

```python
lamps = [Lamp("rood"), DimmableLamp("wit", 30), NightLamp("geel"), ReadingLamp("groen")]
for lamp in lamps:
    lamp.switch()
    print(lamp.describe())
```

## Opdracht 10

`Candle` heeft geen superklasse tussen haakjes.

```python
class Candle:
    """Een kaars: geen lamp, maar je kunt hem wel aan- en uitdoen."""

    def __init__(self):
        """Maak een kaars die nog niet brandt."""
        self.lit = False

    def switch(self):
        """Steekt de kaars aan als hij uit is, en blaast hem uit als hij brandt."""
        self.lit = not self.lit

    def describe(self):
        """Geeft een korte beschrijving, zoals kaars: brandt."""
        if self.lit:
            return "kaars: brandt"
        return "kaars: uit"
```

```python
k = Candle()
assert k.describe() == "kaars: uit"
k.switch()
assert k.describe() == "kaars: brandt"
things = [Lamp("rood"), Candle(), DimmableLamp("wit", 30)]
for thing in things:
    thing.switch()
assert [thing.describe() for thing in things] == [
    "lamp rood: aan",
    "kaars: brandt",
    "lamp wit: aan, 30%",
]
```

## Opdracht 12

De lus drukt af: `Lamp True False`, `DimmableLamp True True`, `ReadingLamp True True`
en `Candle False False`.

1. `ReadingLamp` is een subklasse van `DimmableLamp`, en die van `Lamp`. Een
   leeslamp is dus een lamp. `Candle` erft van niets, en is voor `isinstance` geen
   lamp.
2. De kaars zou niet meer meedoen, terwijl ze alles heeft wat de lus gebruikt.

```python
things = [Lamp("rood"), DimmableLamp("wit", 30), ReadingLamp("groen"), Candle()]
for thing in things:
    print(
        thing.__class__.__name__,
        isinstance(thing, Lamp),
        isinstance(thing, DimmableLamp),
    )
```

## Opdracht 13

`__repr__` staat in `Lamp`, en de subklassen erven haar. `self.__class__.__name__`
is bij elk object de naam van zijn eigen klasse. Hieronder `Lamp` met `__repr__`, en
de twee subklassen opnieuw, zodat ze van deze `Lamp` erven.

```python
class Lamp:
    """Een lamp die aan of uit staat."""

    def __init__(self, color):
        """Maak een lamp met de gegeven kleur; hij staat uit."""
        self.color = color
        self.on = False

    def __repr__(self):
        """Geeft de lamp als string, met de naam van de klasse en de kleur."""
        return f"{self.__class__.__name__}({self.color})"

    def switch(self):
        """Zet de lamp aan als hij uit staat, en uit als hij aan staat."""
        self.on = not self.on

    def describe(self):
        """Geeft een korte beschrijving, zoals lamp rood: uit."""
        if self.on:
            return f"lamp {self.color}: aan"
        return f"lamp {self.color}: uit"


class DimmableLamp(Lamp):
    """Een lamp die je kunt dimmen."""

    def __init__(self, color, level=100):
        """Maak een dimbare lamp; zonder level brandt hij op 100%."""
        super().__init__(color)
        self.level = level

    def describe(self):
        """Geeft de beschrijving van een lamp, met de helderheid erachter."""
        return super().describe() + f", {self.level}%"


class ReadingLamp(DimmableLamp):
    """Een leeslamp: een dimbare lamp die op 60% staat."""

    def __init__(self, color):
        """Maak een leeslamp op 60%."""
        super().__init__(color, 60)
```

```python
assert repr(Lamp("rood")) == "Lamp(rood)"
assert repr(DimmableLamp("wit", 30)) == "DimmableLamp(wit)"
assert repr(ReadingLamp("groen")) == "ReadingLamp(groen)"
```
