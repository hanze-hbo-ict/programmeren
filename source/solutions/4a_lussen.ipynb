## Opdracht 3

1. Verzamelen: `result`, startwaarde `[]`.
2. Langs: de getallen van 1 tot en met `n`.
3. Per stap: voeg het getal toe als `n` deelbaar is door dat getal.
4. Klaar: als alle getallen zijn bekeken.
5. Teruggeven: de lijst met factoren.


```python
def factors(n):
    """Geeft alle factoren van n terug."""
    result = []
    for number in range(1, n + 1):
        if n % number == 0:
            result += [number]
    return result

assert factors(12) == [1, 2, 3, 4, 6, 12]
assert factors(15) == [1, 3, 5, 15]
print(factors(12))
```

## Opdracht 4


```python
def reverse(string):
    result = ""
    for char in string:
        result = char + result
    return result

print(reverse("Hanze"))
```

## Opdracht 5

1. Verzamelen: `result`, startwaarde `0`.
2. Langs: de tekens van `string`.
3. Per stap: verhoog `result` als het teken een klinker is. De operator `in` controleert of een teken in de string met klinkers voorkomt.
4. Klaar: als alle tekens zijn bekeken.
5. Teruggeven: het aantal klinkers.


```python
def count_vowels(string):
    """Telt de klinkers in string."""
    result = 0
    for char in string:
        if char in "aeiou":
            result += 1
    return result

assert count_vowels("Hanze") == 2
assert count_vowels("rhythm") == 0
print(count_vowels("Hanze"))
```
