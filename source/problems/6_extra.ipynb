# Extra

## Van tekst naar beeld

In het werkcollege heb je bestaande afbeeldingen bewerkt. In deze facultatieve opgave maak je zelf een klein beeldbestand uit tekstdata. Je leest `assets/patroon.txt`: een rechthoekig raster met alleen `.` en `#`. Een punt wordt wit en een hekje zwart.

De voorbeeldasset heeft 8 rijen en 12 kolommen. Kies voor je programma een vaste schaalfactor van 8: iedere rastercel wordt een blok van 8 bij 8 pixels. Het resultaat voor `patroon.txt` is dus 96 pixels breed en 64 pixels hoog en heet `patroon.png`.

Gebruik de bestaande `png.py` uit `fraaie_plaatjes.zip` van het werkcollege. Deze opgave bewerkt geen bestaand beeld: ze zet tekstdata om naar een nieuw beeldbestand.

## Geldige invoer

Een geldig raster is niet leeg, heeft geen lege regels, is rechthoekig en bevat uitsluitend `.` en `#`. Schrijf eerst functies die het raster lezen en controleren. Houd de schaalfactor 8 vast; maak die niet afhankelijk van de invoer.

## Wat je gaat maken

Schrijf functies die:

1. `lees_raster(bestandsnaam)` een tekstbestand leest en een lijst van regels teruggeeft;
2. `raster_naar_pixels(raster, schaal)` een lijst van RGB-pixels maakt;
3. `schrijf_patronenbeeld(bestandsnaam, raster)` de pixels met `save_rgb` uit `png.py` opslaat.

Een pixel is een lijst `[rood, groen, blauw]`. Gebruik `[255, 255, 255]` voor wit en `[0, 0, 0]` voor zwart. Test je functies met assertions: controleer onder andere het aantal rijen en kolommen, de RGB-pixels, de vaste schaalfactor en de afmetingen van het uitvoerbeeld. Voer daarna je programma uit op `assets/patroon.txt`.

## Controle

Voor de meegeleverde asset verwacht je 8 rijen, 12 kolommen en een uitvoer van 64 rijen met 96 pixels per rij. Open `patroon.png` om te controleren of de vorm overeenkomt met het tekstbestand. Een niet-rechthoekig raster of een ander teken moet met een assertion worden afgewezen.
