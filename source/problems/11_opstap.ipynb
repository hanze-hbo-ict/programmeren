### Opdracht 4

Deze functie is `exact_change` uit het eerste college, met een `print` aan het
begin. Zo zie je elke aanroep.

a. Hoeveel aanroepen van `function` drukt dit programma af?\
b. Teken de boom van aanroepen, zoals de figuur van `subset` in het college. Zet
in elk vak het bedrag en de munten, en wat de aanroep teruggeeft.\
c. Wat drukt de laatste regel af?

```python
def function(amount, coins):
    print(amount, coins)
    if amount == 0:
        return True
    elif amount < 0:
        return False
    elif coins == []:
        return False
    else:
        use_it = function(amount - coins[0], coins[1:])
        lose_it = function(amount, coins[1:])
        return use_it or lose_it


print(function(3, [1, 2]))
```

### Opdracht 5

Deze functie is `subset` uit het eerste college.

a. Wat geeft `function(5, [4, 3, 2])` terug, en welke items horen daarbij?\
b. Bij welke aanroepen past het eerste item niet, zodat alleen *lose it* wordt
geprobeerd?

```python
def function(capacity, items):
    if items == []:
        return 0
    first = items[0]
    rest = items[1:]
    if first > capacity:
        return function(capacity, rest)
    else:
        use_it = first + function(capacity - first, rest)
        lose_it = function(capacity, rest)
        return max(use_it, lose_it)


print(function(5, [4, 3, 2]))
```
