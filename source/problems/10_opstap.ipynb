### Opdracht 1

a. Wat doet de functie `function`?\
b. Wat drukt dit programma af?

```python
def function(x):
    print(x)
    if x == 0:
        return
    function(x - 1)


function(10)
```

### Opdracht 2

a. Wat doet de functie `function`?\
b. Wat drukt dit programma af?\
c. Wat gebeurt er als je `function(10, 2)` vervangt door `function(5, 0)`?
Voorspel het eerst, en probeer het dan.

```python
def function(x, y):
    print(x)
    if x == 0:
        return
    function(x - y, y)


function(10, 2)
```

### Opdracht 3

a. Wat doet de functie `function`?\
b. Wat drukt dit programma af?

```python
def function(x, y):
    print(x)
    if x >= 20:
        return
    function(x + y, y)


function(10, 6)
```

### Opdracht 4

Hier combineert de functie het antwoord pas ná de recursieve aanroep, net als
`rest = fac(n - 1)` in het college.

a. Wat geeft `function("abc")` terug?\
b. Hoeveel frames van `function` staan er op het diepste punt tegelijk op de
stack?\
c. Wat doet de functie `function`?

```python
def function(s):
    if s == "":
        return ""
    rest = function(s[1:])
    return rest + s[0]


print(function("abc"))
```

### Opdracht 10

Een palindroom is een woord dat achterstevoren hetzelfde is, zoals *lepel* of *kok*. Schrijf de functie `is_palindrome(s)`. Die geeft `True` als `s` een palindroom is, en anders `False`.

| Aanroep | Resultaat |
|---|---|
| `is_palindrome("lepel")` | `True` |
| `is_palindrome("lepels")` | `False` |
| `is_palindrome("")` | `True` |

:::{admonition} Hint
:class: tip

Een string met hoogstens één teken is altijd een palindroom. Dat is dus het basisgeval: `len(s) <= 1`. Is het eerste teken `s[0]` anders dan het laatste, `s[-1]`, dan ben je meteen klaar. Anders kijk je naar wat ertussen staat: `s[1:-1]`.
:::

```python
# jouw oplossing
```

```python
assert is_palindrome("lepel") == True
assert is_palindrome("kok") == True
assert is_palindrome("lepels") == False
assert is_palindrome("ab") == False
assert is_palindrome("a") == True
assert is_palindrome("") == True
```

### Opdracht 11

Twee woorden rijmen in deze opdracht op `n` letters als hun laatste `n` letters
gelijk zijn. Zo rijmen *kater* en *water* op 4 letters: allebei eindigen ze op
*ater*. Dat kun je recursief bepalen, vanaf het eind:

- Is `n` gelijk aan 0, dan geef je `True` terug.
- Is een van de twee woorden leeg, dan geef je `False` terug.
- Is de laatste letter van het ene woord niet gelijk aan de laatste letter van
  het andere, dan geef je `False` terug.
- Anders roep je de functie opnieuw aan, met beide woorden zonder hun laatste
  letter en met `n - 1`, en geef je het resultaat daarvan terug.

Schrijf de functie `rhymes(word1, word2, n)` volgens deze regels.

| Aanroep | Resultaat |
|---|---|
| `rhymes("kater", "water", 4)` | `True` |
| `rhymes("kat", "kip", 1)` | `False` |
| `rhymes("aap", "schaap", 4)` | `False` |

:::{admonition} Hint
:class: tip

Elke regel wordt één tak van je `if`, in dezelfde volgorde. De laatste letter van
`s` is `s[-1]`, en alles ervóór is `s[:-1]`.
:::

```python
# jouw oplossing
```

```python
assert rhymes("kater", "water", 4) == True
assert rhymes("kat", "rat", 2) == True
assert rhymes("kat", "kip", 1) == False
assert rhymes("aap", "schaap", 3) == True
assert rhymes("aap", "schaap", 4) == False
```
