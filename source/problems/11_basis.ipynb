## De nacht van het nijlpaard

Nijlpaarden houden van een goede maaltijd, maar ze willen wel ruimte om zich
heen. Elke nacht zet de kok maaltijden op een lange tafel. Aan één kant van de
tafel zijn de plaatsen genummerd: 1, 2, 3, enzovoort. Op een plaats kan meer dan
één maaltijd staan. De ceremoniemeester bepaalt hoeveel ruimte er tussen de
nijlpaarden moet zijn, en zorgt dat er zoveel mogelijk nijlpaarden mee-eten. In
deze opgave reken je uit hoeveel dat er zijn.

### De regel

Een nijlpaard gaat zitten op een plaats waar een maaltijd staat. De **afstand**
tussen twee nijlpaarden is het verschil van hun plaatsnummers, en die moet
minstens gelijk zijn aan de afstand die de ceremoniemeester heeft gekozen. Op één
plaats zit hoogstens één nijlpaard.

| Gekozen afstand | Plaatsen met een maaltijd | Zoveel nijlpaarden eten mee | Bijvoorbeeld op |
|---|---|---|---|
| 1 | 2, 3, 3 | 2 | 2 en 3 |
| 2 | 1, 3, 7 | 3 | 1, 3 en 7 |
| 4 | 1, 3, 5, 10, 12 | 3 | 1, 5 en 12 |

Bij afstand 2 zit er dus minstens één lege plaats tussen twee nijlpaarden. In het
laatste voorbeeld kunnen er geen vier mee-eten: welke vier plaatsen je ook kiest,
er liggen er altijd twee minder dan 4 uit elkaar.

![Voorbeeld 1: maaltijden op plaats 2, 3 en 3; nijlpaarden zitten op plaats 2 en 3](images/11/example1.png)

![Voorbeeld 2: maaltijden op plaats 1, 3 en 7; op alle drie zit een nijlpaard](images/11/example2.png)

![Voorbeeld 3: maaltijden op plaats 1, 3, 5, 10 en 12; nijlpaarden zitten op plaats 1, 5 en 12](images/11/example3.png)

De plaatsen staan in deze opgave altijd oplopend in de lijst.

### Wat je gaat maken

| Stap | Functie | Doet | Wat je oefent |

| 3 | `hippo_dinner` | hoeveel nijlpaarden kunnen er hoogstens mee-eten? | *use it or lose it* |

### Stap 1: `far_enough(a, b, distance)`

Geeft `True` als plaats `b` minstens `distance` verder ligt dan plaats `a`, en
anders `False`. `b` is nooit kleiner dan `a`. Deze functie heeft geen recursie
nodig.

| Aanroep | Resultaat |
|---|---|
| `far_enough(1, 3, 2)` | `True` |
| `far_enough(2, 3, 2)` | `False` |
| `far_enough(3, 3, 1)` | `False` |

:::{admonition} Hint
:class: tip

Het verschil tussen de twee plaatsen is `b - a`.
:::

```python
# jouw oplossing
```

```python
assert far_enough(1, 3, 2) == True
assert far_enough(2, 3, 2) == False
assert far_enough(3, 3, 1) == False
assert far_enough(5, 10, 5) == True
```

### Stap 2: `drop_too_close(place, places, distance)`

Er zit een nijlpaard op plaats `place`. `places` is een oplopende lijst met de
plaatsen erna. Geeft de lijst terug zonder de plaatsen aan het begin die te dicht
bij `place` liggen. Vanaf de eerste plaats die ver genoeg ligt, blijft de lijst
staan.

| Aanroep | Resultaat |
|---|---|
| `drop_too_close(1, [3, 7], 2)` | `[3, 7]` |
| `drop_too_close(3, [3, 5, 8], 4)` | `[8]` |
| `drop_too_close(5, [], 2)` | `[]` |

:::{admonition} Hint
:class: tip

Gebruik `far_enough` uit stap 1. Ligt de eerste plaats ver genoeg, dan ben je
klaar: dan geef je de hele lijst terug. Omdat de lijst oplopend is, liggen de
plaatsen daarna nog verder weg.
:::

```python
# jouw oplossing
```

```python
assert drop_too_close(1, [3, 7], 2) == [3, 7]
assert drop_too_close(3, [3, 5, 8], 4) == [8]
assert drop_too_close(1, [3, 5, 10, 12], 4) == [5, 10, 12]
assert drop_too_close(5, [6, 7], 3) == []
assert drop_too_close(5, [], 2) == []
```

### Stap 3: `hippo_dinner(distance, places)`

`places` is een oplopende lijst met de plaatsen waar een maaltijd staat. Geeft
terug hoeveel nijlpaarden er hoogstens kunnen mee-eten als de afstand tussen
twee nijlpaarden minstens `distance` moet zijn. Gebruik recursie.

| Aanroep | Resultaat |
|---|---|
| `hippo_dinner(1, [2, 3, 3])` | `2` |
| `hippo_dinner(2, [1, 3, 7])` | `3` |
| `hippo_dinner(4, [1, 3, 5, 10, 12])` | `3` |
| `hippo_dinner(3, [])` | `0` |

:::{admonition} Hint
:class: tip

Kies bij de eerste plaats, zoals bij `subset` in het
[college](/lectures/11a_use_it_or_lose_it):

- *Use it*: er gaat een nijlpaard op de eerste plaats zitten. Dat is er één, plus
  zoveel als er passen op de rest, nadat je met `drop_too_close` de plaatsen hebt
  weggehaald die te dicht bij de eerste liggen.
- *Lose it*: op de eerste plaats gaat niemand zitten. Dan tel je zoveel als er
  op de rest passen.

Geef het grootste van de twee terug.
:::

```python
# jouw oplossing
```

```python
assert hippo_dinner(1, [2, 3, 3]) == 2
assert hippo_dinner(2, [1, 3, 7]) == 3
assert hippo_dinner(4, [1, 3, 5, 10, 12]) == 3
assert hippo_dinner(3, [4]) == 1
assert hippo_dinner(3, []) == 0
```

### Tot slot

Bij afstand 4 en maaltijden op 1, 3, 5, 10 en 12 eten er drie nijlpaarden mee.
Of je de eerste plaats moet gebruiken, kon `hippo_dinner` niet vooraf weten. Pas
door beide keuzes te proberen vond de functie het beste antwoord, precies zoals
`subset` in het college.

`hippo_dinner` zegt hoeveel nijlpaarden er mee-eten, maar niet op welke plaatsen.

## De uitslag van de schoolrun

Op de sportdag lopen de leerlingen een rondje van drie kilometer. Na afloop moet

### De uitslag

De uitslag is een lijst van lijsten. Elke deelnemer is een lijst met drie
elementen:

| Index | Betekenis | Voorbeeld |
|---|---|---|
| `0` | naam | `"Noor"` |
| `1` | leeftijd in jaren | `17` |

Noor liep het rondje dus in 754 seconden: 12 minuten en 34 seconden. Wie de

```python
RESULTS = [
    ["Noor", 17, 754],
    ["Sem", 19, 698],
    ["Lotte", 18, 698],
    ["Yara", 16, 812],
    ["Finn", 21, 745],
]
```

### Stap 4: `fastest(results)`

Geeft de naam van de deelnemer met de kleinste tijd. `results` is niet leeg.
Hebben twee deelnemers dezelfde tijd, dan wint wie het eerst in de lijst staat.

| Aanroep | Resultaat |
|---|---|
| `fastest(RESULTS)` | `"Sem"` |
| `fastest([["Ali", 20, 600]])` | `"Ali"` |

:::{admonition} Hint
:class: tip

`min` met een lambda-functie achter `key=` geeft de hele deelnemer met de
kleinste tijd, zoals `max(players, key=lambda p: p[1])` in het
[tweede college](/lectures/11b_functie_als_argument). Van die deelnemer geef je
alleen de naam terug. Bij gelijke tijden geeft `min` vanzelf de eerste.
:::

```python
# jouw oplossing
```

```python
assert fastest(RESULTS) == "Sem"
assert fastest([["Ali", 20, 600]]) == "Ali"
```

### Stap 5: `ranking(results)`

Geeft een lijst met de namen van alle deelnemers, van de snelste naar de
langzaamste. Hebben twee deelnemers dezelfde tijd, dan komt de naam die in het
alfabet eerst komt vooraan.

| Aanroep | Resultaat |
|---|---|
| `ranking(RESULTS)` | `["Lotte", "Sem", "Finn", "Noor", "Yara"]` |
| `ranking([])` | `[]` |

Sem en Lotte hebben allebei 698 seconden. Lotte staat vooraan, want de *L* komt
vóór de *S*.

:::{admonition} Hint
:class: tip

Een sleutel mag ook een lijst zijn. Python vergelijkt twee lijsten zoals twee
woorden: eerst het eerste element, en pas als dat gelijk is het tweede.
`[698, "Lotte"]` komt dus vóór `[698, "Sem"]`. Laat de lambda-functie de lijst
`[tijd, naam]` teruggeven. Zet daarna met een list comprehension alleen de namen
in een lijst.
:::

```python
# jouw oplossing
```

```python
assert ranking(RESULTS) == ["Lotte", "Sem", "Finn", "Noor", "Yara"]
assert ranking([["Bo", 30, 700], ["Ad", 25, 700]]) == ["Ad", "Bo"]
assert ranking([]) == []
```

### Stap 6: `select(results, test)`

`test` is een functie die bij een deelnemer `True` of `False` teruggeeft. Geeft
een lijst met de deelnemers uit `results` waarvoor `test` `True` teruggeeft, in
dezelfde volgorde.

| Aanroep | Resultaat |
|---|---|
| `select(RESULTS, lambda r: r[1] < 18)` | `[["Noor", 17, 754], ["Yara", 16, 812]]` |
| `select(RESULTS, lambda r: r[1] > 30)` | `[]` |

De lambda-functie in het eerste voorbeeld geeft `True` voor wie jonger is dan
18: de jeugd.

:::{admonition} Hint
:class: tip

Een list comprehension met een `if`, zoals in week 2. Roep in de `if` de functie
`test` aan op elke deelnemer.
:::

```python
# jouw oplossing
```

```python
assert select(RESULTS, lambda r: r[1] < 18) == [["Noor", 17, 754], ["Yara", 16, 812]]
assert select(RESULTS, lambda r: r[2] < 700) == [["Sem", 19, 698], ["Lotte", 18, 698]]
assert select(RESULTS, lambda r: r[1] > 30) == []
```

### Opdracht 1: dezelfde regel bij gelijke tijden

Maak eerst stap 4 tot en met 8 af. Voer de aparte analyseversie hieronder uit, ook als jouw
eigen code anders is. Ze vervangt de gelijknamige functies en `RESULTS` in je
notebook, maar bevat geen uitwerking van het jeugdpodium.

De school wil voortaan ook bij `fastest` en `winner` bij gelijke tijden de
alfabetisch eerste naam als winnaar. Je beschrijft de wijziging; je hoeft haar
niet te programmeren. De voorwaarden voor niet-lege invoer blijven gelden.


```python
RESULTS = [
    ["Noor", 17, 754],
    ["Sem", 19, 698],
    ["Lotte", 18, 698],
    ["Yara", 16, 812],
    ["Finn", 21, 745],
]

def fastest(results):
    """Geeft de naam van de deelnemer met de kleinste tijd; results is niet leeg."""
    return min(results, key=lambda r: r[2])[0]


assert fastest(RESULTS) == "Sem"
assert fastest([["Ali", 20, 600]]) == "Ali"

def ranking(results):
    """Geeft de namen van snel naar langzaam; bij gelijke tijd op alfabet."""
    return [r[0] for r in sorted(results, key=lambda r: [r[2], r[0]])]


assert ranking(RESULTS) == ["Lotte", "Sem", "Finn", "Noor", "Yara"]
assert ranking([["Bo", 30, 700], ["Ad", 25, 700]]) == ["Ad", "Bo"]
assert ranking([]) == []

def select(results, test):
    """Geeft de deelnemers uit results waarvoor test True teruggeeft."""
    return [r for r in results if test(r)]


assert select(RESULTS, lambda r: r[1] < 18) == [["Noor", 17, 754], ["Yara", 16, 812]]
assert select(RESULTS, lambda r: r[2] < 700) == [["Sem", 19, 698], ["Lotte", 18, 698]]
assert select(RESULTS, lambda r: r[1] > 30) == []

def winner(results, test):
    """Geeft de naam van de snelste deelnemer waarvoor test True teruggeeft."""
    return fastest(select(results, test))


assert winner(RESULTS, lambda r: r[1] < 18) == "Noor"
assert winner(RESULTS, lambda r: r[1] >= 18) == "Sem"
assert winner(RESULTS, lambda r: r[0][0] == "Y") == "Yara"
```
