De opdrachten bouwen voort op de klasse `Lamp` uit de opstap van week 5, met één
methode erbij: `describe`, die een korte beschrijving teruggeeft. Voer de cel
hieronder eerst uit.

````python
class Lamp:
    """Een lamp die aan of uit staat."""

    def __init__(self, color):
        """Maak een lamp met de gegeven kleur; hij staat uit."""
        self.color = color
        self.on = False

    def switch(self):
        """Zet de lamp aan als hij uit staat, en uit als hij aan staat."""
        self.on = not self.on

    def describe(self):
        """Geeft een korte beschrijving, zoals lamp rood: uit."""
        if self.on:
            return f"lamp {self.color}: aan"
        return f"lamp {self.color}: uit"
````

Schrijf de tekst van `Lamp` niet opnieuw. `super().describe()` roept de methode
