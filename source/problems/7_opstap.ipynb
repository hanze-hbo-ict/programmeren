# Opstap

Werk met een rechthoekig raster: een lijst van lijsten met `0` (dood) en `1` (levend). `bord[rij][kolom]` is één cel. Elke opdracht staat op zichzelf.

## 1. Eén cel veranderen

Maak dit bord en verander alleen de cel op rij 1, kolom 2 in `1`. Controleer dat de andere cellen hetzelfde blijven.

```python
bord = [[0, 0, 0], [0, 0, 0], [0, 0, 0]]
# schrijf hier de directe toekenning
assert bord == [[0, 0, 0], [0, 0, 1], [0, 0, 0]]
```

## 2. Aliasing

Voer `ander = bord` uit. Verander daarna `ander[0][0]` in `1` en leg uit waarom `bord` ook verandert.

```python
bord = [[0, 0], [0, 0]]
ander = bord
# verander ander hier
assert bord[0][0] == 1
```

## 3. Een onafhankelijke kopie

Maak `kopie` met een nieuwe buitenlijst en voor elke rij een nieuwe lijst. Verander daarna alleen `kopie[0][0]`.

```python
bord = [[0, 0], [0, 0]]
kopie = []
for rij in bord:
    kopie = kopie + [rij[:]]
kopie[0][0] = 1
assert bord == [[0, 0], [0, 0]]
assert kopie == [[1, 0], [0, 0]]
```

## 4. Tuple en uitpakken

Schrijf `midden_van` zodat zij de positie van het midden teruggeeft. Pak de twee waarden daarna uit.

```python
def midden_van(bord):
    return (len(bord) // 2, len(bord[0]) // 2)
positie = midden_van([[0, 0, 0], [0, 1, 0], [0, 0, 0]])
rij, kolom = positie
assert (rij, kolom) == (1, 1)
```
